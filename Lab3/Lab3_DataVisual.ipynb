# **Zadanie 1 (Wykres funkcji matematycznej, stylizacja linii)**
W uczeniu maszynowym często wizualizujemy *funkcje kosztu* lub *funkcje aktywacji (np. Sigmoid, ReLU)*. Twoim zadaniem jest zwizualizowanie nieliniowej funkcji matematycznej.
1. Za pomocą *np.linspace* stwórz wektor *x* zawierający 100 punktów w przedziale od \([-5, 5]\).
2. Oblicz wartości funkcji y = x² - 3x + 2.
3. Za pomocą *matplotlib.pyplot* narysuj wykres funkcji kwadratowej przerywaną linią w kolorze czerwonym o grubości 2. Dodaj siatkę pomocniczą (plt.grid), tytuł wykresu oraz etykiety osi X i Y.
4. Modyfikować kod, żeby otrzymać wykres funkcji sigmoidalnej (y = 1/(1+e^x)), kolor niebieski, ciągła linia.
5. Omówić własność **funkcji sigmoidalnej**:

  *(dziedzina, przeciwdziedzina, asymptota, miejsce zerowe, punkt przecięcia z osią Oy, monotoniczność, rożniczkowalność, wzór funkcji pochodnej)*.  

```python
import numpy as np
import matplotlib.pyplot as plt
# Twój kod tutaj
x = np.linspace(-5,5,100)
y = 1/(1+np.e**x)

#Z jakiegoś powodu w moim edytorze wykresy domyślnie mają odwrócone kolory, to powinno ustawić je znowu na białe (jak w Colabie), jak wygląda nieczytelnie w Colabie to przepraszam!
plt.style.use('default')
plt.rcParams.update({
    'figure.facecolor': 'white',
    'axes.facecolor':   'white',
    'savefig.facecolor':'white',
    'text.color':       'black',
    'axes.labelcolor':  'black',
    'axes.edgecolor':   'black',
    'xtick.color':      'black',
    'ytick.color':      'black',
})

plt.plot(x, y, 'b', linewidth=2)
plt.xlabel('x')
plt.ylabel('y')
plt.title("Wykres funkcji kwadratowej")
plt.grid(True)
plt.title("Zadanie 1")

print(f"\nKOMENTARZ DO 1.5\n\nDziedzina: R (a tu akurat [-5;5])\nPrzeciwdziedzina: (0;1)\nAsymptoty: y=0 i y=1")
print(f"Miejsce zerowe: brak, zawsze y>0\nPrzecięcie z Oy: y(0) = 0.5\nMonotnicznośc: Ściśle malejąca w całym przedziale")
print(f"Różnicznowalność: w każdym punkcie, nieskończenie wiele razy\nWzór pochodnej: (e**x)/((1+e**x)**2)")
plt.show()
```


## **Zadanie 2: Kilka wykresów i funkcje trygonometryczne (Subplots)**

Często zachodzi potrzeba porównania kilku rozkładów lub przebiegów procesów na w jednym panelu (obok siebie).

1. Stwórz wektor *t* od 0 do 4π z krokiem 0.05.
2. Korzystając z *plt.subplots*, stwórz panel składający się z dwóch wykresów obok siebie (1 wiersz, 2 kolumny) o rozmiarze całego okna (14, 5).
3. Na pierwszym (lewym) wykresie narysuj funkcję \(\sin(t)\) (kolor niebieski).
4. Na drugim (prawym) wykresie narysuj funkcję \(\cos(t)\) (kolor zielony).
5. Upewnij się, że każdy sub-wykres ma swój własny tytuł, a opisy osi nie nakładają się na siebie (plt.tight_layout()).



```python
import numpy as np
import matplotlib.pyplot as plt

t = np.arange(0,4*np.pi,0.05)

fig, (ax1,ax2) = plt.subplots(1, 2, figsize=(14,5))

fig.suptitle('Wykresy')
ax1.plot(t, np.sin(t), 'b-', linewidth=2)
ax2.plot(t, np.cos(t), 'g-', linewidth=2)
ax1.set_title('Sin(t)')
ax2.set_title('Cos(t)')
ax1.set_xlabel('t')
ax2.set_xlabel('t')
ax1.set_ylabel('sin(t)')
ax2.set_ylabel('cos(t)')
ax1.grid(True)
ax2.grid(True)
plt.tight_layout()
plt.show()
```

## **Zadanie 6 (Dwa wykresów w jednym układzie współrzędnych)**

Podczas analizy algorytmów optymalizacyjnych często porównujemy tempo zbieżności różnych funkcji kosztu. Twoim zadaniem jest nałożenie na siebie dwóch przebiegów matematycznych, aby naocznie ocenić ich zachowanie w tym samym przedziale dziedziny.
1. Za pomocą *np.linspace* wygeneruj wektor x składający się z 250 punktów w przedziale od \([-2, 4]\).
2. Oblicz wartości dla dwóch niezależnych funkcji:
*   Funkcja A (parabola): $y_1 = x^2 - 2x$
*   Funkcja B (funkcja sześcienna): $y_2 = 0.5x^3 - x$

3. Narysuj oba wykresy w obrębie tego samego, pojedynczego układu współrzędnych (wywołując dwukrotnie funkcję plt.plot przed wyświetleniem okna):

*   Przebieg Funkcji A oznacz kolorem niebieskim, linią ciągłą o grubości 2.
*  Przebieg Funkcji B oznacz kolorem pomarańczowym, linią przerywaną (linestyle='--') o grubość 2.

4. Konfiguracja czytelności (Wymagane):

  * Dodaj etykiety tekstowe do obu linii i wyświetl na wykresie legendę (plt.legend()) w lewym górnym rogu.
  * Włącz siatkę pomocniczą (plt.grid())
  * Nadaj tytuł wykresowi oraz opisz osie X i Y.



```python
x = np.linspace(-2,4,250)
funcA = x**2 - 2*x
funcB = 0.5*x**3 - x

plt.plot(funcA,color='b',linewidth=2,label='A',linestyle='-')
plt.plot(funcB,color='orange',linewidth=2,label='B',linestyle='--')
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.title("Wykres funkcji A i B")
plt.show()
```
